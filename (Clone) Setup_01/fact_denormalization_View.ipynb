```python
%sql
CREATE OR REPLACE VIEW ebook.gold.fact_transaction_denormalized As(
    SELECT i.*,
    c.year,
    c.month_name,
    c.is_weekend,
    c.quarter,
    c.week,
    p.sku,
    p.category_code,
    p.category_name,
    p.brand_code,
    p.color,
    p.size,
    p.rating_count,
    extract(Hour from transaction_ts) as hour_of_day
    FROM ebook.gold.gld_fact_order_items i join ebook.gold.gld_dim_date c ON i.date_id = c.date_id 
    join ebook.gold.gld_dim_products p ON i.product_id = p.product_id
);
```
